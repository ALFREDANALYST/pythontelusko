### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
## Write your code here
df.isna().sum()
```

```python
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
```

```python
# Columns to impute
cols_to_impute = ['JobSatPoints_11', 
                   'ConvertedCompYearly', 'JobSat', 'YearsCodePro']

for col in cols_to_impute:
    if col in df.columns:
        median_val = df[col].median()
        df[col] = df[col].fillna(median_val)
        print(f"{col}: filled with median = {median_val}")
    else:
        print(f"{col}: not found in dataset")

# Verify no missing values remain in these columns
print(df[cols_to_impute].isnull().sum())
```

```python
df.isna().sum()
```

```python
# Drop any column that has at least one null value
df_cleaned = df.dropna(axis=0)
```

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
## Write your code here
plt.figure(figsize=(8, 5))
sns.kdeplot(df_cleaned['JobSat'].dropna(), fill=True, color='steelblue')
plt.title('Distribution of JobSat (KDE)')
plt.xlabel('JobSat')
plt.ylabel('Density')
plt.tight_layout()
plt.show()
```
